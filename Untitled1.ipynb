```python
import tkinter as tk
from tkinter import messagebox

# Kisi sınıfı, kişisel bilgileri tutar ve bunları yönetir
class Kisi:
    def __init__(self, isim, soyisim, yas, boy, ülke, email, telefon, ilgi_alanları): # girilmesi gereken isim,soyisim,yas vesaire parametreleri belirledim
        self.isim = isim
        self.soyisim = soyisim
        self.yaş = yaş
        self.boy = boy
        self.ülke = ülke
        self.email = email
        self.telefon = telefon
        self.ilgi_alanları = ilgi_alanları

    # Kişi bilgilerini gösteren fonksiyonu oluşturdum
    def bilgileri_göster(self):
        return f"İsim: {self.isim}\nSoyisim: {self.soyisim}\nYaş: {self.yaş}\nBoy: {self.boy}\nÜlke: {self.ülke}\nEmail: {self.email}\nTelefon: {self.telefon}\nİlgi Alanları: {', '.join(self.ilgi_alanları)}"
         #burada kullanıcıda olması gereken özellikleri ekrana yazdırmak için bilgileri_göster fonksiyonu ekledim

   # Kişi bilgilerini güncelleyen fonksiyonu oluşturdum
    def bilgileri_güncelle(self, isim, soyisim, yaş, boy, ülke, email, telefon, ilgi_alanları):
        self.isim = isim
        self.soyisim = soyisim
        self.yaş = yaş
        self.boy = boy
        self.ülke = ülke
        self.email = email
        self.telefon = telefon
        self.ilgi_alanları = ilgi_alanları


# GUI (Grafiksel Kullanıcı Arayüzü) sınıfı, tkinter kullanarak arayüzü oluşturur
class GUI(tk.Tk):
    def __init__(self):
        super().__init__() # daha önce belirlediğim parametreleri getirdim super().init ile
        self.title("--Bilgi Kayıt Ekranı--")  # Kayıt Ekranı başlığını belirledim
        self.geometry("400x600")  # Pencere boyutunu ayarladım(400x600)

        self.iconbitmap("ikonum.ico")  # Uygulama simgesini ayarladım

        self.kisi = None  # Başlangıçta kişi bilgileri boş olarak belirledim

        self.create_widgets()  # Bileşenleri oluşturmak için metodu çağırdım

    def create_widgets(self):
        font_settings = ("Times", 12, "bold")#yazı tipini ve kalınlığını ayarladım

        # İsim bilgisi alanı oluşturdum
        self.isim_label = tk.Label(self, text="İsim Bilgilerinizi Giriniz:", fg="blue", font=font_settings) #isim bilgisi istedik,renk ve yazı boyutunu ayarladım
        self.isim_label.pack()  # Label'ı pencereye ekledik
        self.isim_var = tk.StringVar()  # İsim için bir değişken oluşturuyoruz
        self.isim_entry = tk.Entry(self, textvariable=self.isim_var, font=font_settings)  # Kullanıcıdan veri almak için oluşturdum ve font boyutunu ekledim
        self.isim_entry.pack()  # Entry widget'ını pencereye ekledim

        # Soyisim bilgisi alanı oluşturdum
        self.soyisim_label = tk.Label(self, text="Soyisim Bilgilerinizi Giriniz:", fg="purple", font=font_settings)#soyisim bilgisi istedik,renk ve yazı boyutunu ayarladım
        self.soyisim_label.pack()
        self.soyisim_var = tk.StringVar()
        self.soyisim_entry = tk.Entry(self, textvariable=self.soyisim_var, font=font_settings)
        self.soyisim_entry.pack()

        # Yaş bilgisi alanı oluşturdum
        self.yaş_label = tk.Label(self, text="Yaş Bilgilerinizi Giriniz:", fg="black", font=font_settings)#Yaş bilgisi istedik,renk ve yazı boyutunu ayarladım
        self.yaş_label.pack()
        self.yaş_var = tk.IntVar()  # Yaş, bir tamsayı olduğu için IntVar() kullanıyoruz
        self.yaş_entry = tk.Entry(self, textvariable=self.yaş_var, font=font_settings)
        self.yaş_entry.pack()

        # Boy bilgisi alanı oluşturdum
        self.boy_label = tk.Label(self, text="Boy Bilgilerinizi Giriniz:", fg="red", font=font_settings)#Boy bilgisi istedik,renk ve yazı boyutunu ayarladım
        self.boy_label.pack()
        self.boy_var = tk.DoubleVar()  # Boy, bir ondalıklı sayı olduğu için DoubleVar() kullanıyoruz int değer atamak mantıksız olacagından dolayı
        self.boy_entry = tk.Entry(self, textvariable=self.boy_var, font=font_settings) # boy değerini ve font ayarlarını atadık
        self.boy_entry.pack()

        # Ülke bilgisi alanı oluşturdum
        self.ülke_label = tk.Label(self, text="Ülke Bilgilerinizi Giriniz:", fg="darkblue", font=font_settings) #ülke bilgisi istedik,renk ve yazı boyutunu ayarladım
        self.ülke_label.pack()
        self.ülke_var = tk.StringVar() #string ifade olacağından dolayı strink ifade olacağını belirliyoruz
        self.ülke_entry = tk.Entry(self, textvariable=self.ülke_var, font=font_settings) #font ayarlarını ve ülke değerlerini atadık
        self.ülke_entry.pack()

        # Email bilgisi alanı oluşturdum
        self.email_label = tk.Label(self, text="Email Bilgilerini Giriniz:", fg="black", font=font_settings)#email bilgisi istedik,renk ve yazı boyutunu ayarladık
        self.email_label.pack()
        self.email_var = tk.StringVar()
        self.email_entry = tk.Entry(self, textvariable=self.email_var, font=font_settings)
        self.email_entry.pack()

        # Telefon bilgisi alanı oluşturdum
        self.telefon_label = tk.Label(self, text="Telefon Numarası Giriniz:", fg="orange", font=font_settings)#telno bilgisi istedik,renk ve yazı boyutunu ayarladık
        self.telefon_label.pack()
        self.telefon_var = tk.StringVar()
        self.telefon_entry = tk.Entry(self, textvariable=self.telefon_var, font=font_settings)
        self.telefon_entry.pack()

        # İlgi alanları bilgisi alanı oluşturdum
        self.ilgi_alanları_label = tk.Label(self, text="İlgi Alanları giriniz (virgülle ayırılması lazım):", fg="green", font=font_settings)#ilgi alanları bilgisini istedik ve yazı tipi ve renk atamalarını gerçekleştirdim
        self.ilgi_alanları_label.pack()
        self.ilgi_alanları_var = tk.StringVar()
        self.ilgi_alanları_entry = tk.Entry(self, textvariable=self.ilgi_alanları_var, font=font_settings)
        self.ilgi_alanları_entry.pack()

        # Bilgileri güncellemek için buton oluşturdum
        self.update_button = tk.Button(self, text="Bilgileri Güncelle", command=self.bilgileri_güncelle, fg="blue", font=font_settings)
        self.update_button.pack()

        # Bilgileri göstermek için buton oluşturdum
        self.show_button = tk.Button(self, text="Bilgileri Göster", command=self.bilgileri_göster, fg="blue", font=font_settings)
        self.show_button.pack()

    # Bilgileri güncelleme metodu
    def bilgileri_güncelle(self):
        # Kullanıcıdan alınan bilgileri değişkenlere atıyoruz
        isim = self.isim_var.get()
        soyisim = self.soyisim_var.get()
        yaş = self.yaş_var.get()
        boy = self.boy_var.get()
        ülke = self.ülke_var.get()
        email = self.email_var.get()
        telefon = self.telefon_var.get()
        ilgi_alanları = self.ilgi_alanları_var.get().split(", ")  # İlgi alanları virgülle ayrılması gerektiğinden dolayı split ekledik.

        # Eğer kullanıcı kişi nesnesi oluşturmamışsa yeni bir kişi nesnesi oluşturuyorum
        if self.kisi is None: #kişi değerini none olarak belirliyorum daha sonra gelen parametreleri atamak için
            self.kisi = Kisi(isim, soyisim, yaş, boy, ülke, email, telefon, ilgi_alanları) #girilen parametreleri none degeri olan kişiye atıyorum
            messagebox.showinfo("Başarılı", "-Yeni bir kişi başarıyla oluşturuldu-") # yeni kişi oluşturuldu mesajını yazdırdık.
        else:
            # Kişi zaten oluşturulmuşsa bilgileri güncelleniyor
            self.kisi.bilgileri_güncelle(isim, soyisim, yaş, boy, ülke, email, telefon, ilgi_alanları)# gerekli parametreler ekleniyor
            messagebox.showinfo("Başarılı", "-Bilgiler başarıyla güncellendi-") # bilgiler başarıyla güncellendi mesajı ekrana  bastırılıyor.

    # Bilgileri gösterme fonksiyonu
    def bilgileri_göster(self):
        if self.kisi:
            # Kişi bilgileri daha önceden eklenmişse bilgileri gösterir
            info = self.kisi.bilgileri_göster()
            messagebox.showinfo("--Kişi Bilgileri--", info)
        else:
            # Eğer kişi bilgisi hiç girilmemişse kullanıcıyı uyarır
            messagebox.showwarning("Hata", "Önce kişisel bilgileri giriniz ve kayıt ediniz efendim") #uyarı mesajını ekrana bastırdık.


# Ana program çalıştırıldığında GUI başlatılır
if __name__ == "__main__":
    app = GUI()  # GUI sınıfından bir nesne oluşturur e
    app.mainloop()  # Ana döngüyü başlatır
```
